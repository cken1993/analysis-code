```r
limma_pipeline = function(df,exp_tmp){
    group = df$new_col
    design <- model.matrix(~0+group)
    colnames(design) <- c("down","up")

    fit <- lmFit(exp_tmp, design = design)
    contrastsMatrix <- makeContrasts(up - down, levels=design)

    fit2 <- contrasts.fit(fit, contrasts=contrastsMatrix)
    fit2 <- eBayes(fit2, 0.01)
    tT <- topTable(fit2, adjust="fdr", sort.by="B", number=100000)

    tT <- subset(tT, select=c("logFC","t","B","P.Value","adj.P.Val"))
    colnames(tT) <- c("log2FC","t","B","PValue","FDR")
    tT <- tT[order(tT$FDR),]

    return(tT)
}
```

```r
deg_res = list()

df = sig_GSE54236 %>% filter(new_col == 'down' | new_col == 'up')
df$new_col = factor(df$new_col,levels = c('down','up'))
exp_tmp = GSE54236_df[,df$ID]
dim(exp_tmp)
deg_res[['GSE54236']] = limma_pipeline(df,exp_tmp)

df = sig_GSE64041 %>% filter(new_col == 'down' | new_col == 'up')
df$new_col = factor(df$new_col,levels = c('down','up'))
exp_tmp = GSE64041_df[,df$ID]
dim(exp_tmp)
deg_res[['GSE64041']] = limma_pipeline(df,exp_tmp)

df = sig_GSE76427 %>% filter(new_col == 'down' | new_col == 'up')
df$new_col = factor(df$new_col,levels = c('down','up'))
exp_tmp = GSE76427_df[,df$ID]
dim(exp_tmp)
deg_res[['GSE76427']] = limma_pipeline(df,exp_tmp)
```

```r
library(dplyr)
library(stats)
df = df_tls_GSE76427
genes <- colnames(df)[2:13]
results <- data.frame(Gene = character(), Median_Tumor = numeric(), Median_Control = numeric(), Ratio = numeric(), P_value = numeric(), stringsAsFactors = FALSE)
for (gene in genes) {
  df_gene <- df %>% select(TYPE, !!sym(gene))
  df_gene$TYPE = as.factor(df_gene$TYPE)
  df_gene[gene] = as.numeric(as.numeric(df_gene[,gene]))
  df_grouped <- df_gene %>%
    group_by(TYPE) %>%
    summarize(median_gene = median(!!sym(gene)))
  p_value <- wilcox.test(df_gene[df_gene$TYPE == "HCC",][[2]], df_gene[df_gene$TYPE == "Adjacent",][[2]])$p.value
  result <- data.frame(Gene = gene, 
    Median_Tumor = df_grouped[df_grouped[[1]] == 'HCC',][[2]], 
    Median_Control = df_grouped[df_grouped[[1]] == 'Adjacent',][[2]], 
    Ratio = df_grouped[df_grouped[[1]] == 'HCC',][[2]]/df_grouped[df_grouped[[1]] == 'Adjacent',][[2]], 
    P_value = p_value)
  results <- rbind(results, result)
}
```
